## PROBLEM 1

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("board2 (1).xlsx", sheet_name="board2.csv")
```

```python
df["Average"] = df[["Math", "Electronics", "GEAS", "Communication"]].mean(axis=1)
```

```python
df.head()
```

```python
VisComm = df[(df["Hometown"] == "Visayas") & (df["Track"] == "Communication")][
    ["Name", "Gender", "Math", "Electronics", "Average"]
]

display(VisComm)
print("Number of rows:", VisComm.shape[0])
```

## PROBLEM 2

```python
VisFemale = df[(df["Hometown"] == "Visayas") & (df["Gender"] == "Female")][
    ["Name", "Track", "GEAS", "Electronics", "Average"]
]

display(VisFemale)
```

```python
VisFemale_high = VisFemale[VisFemale["Average"] >= 60]
display(VisFemale_high)
```

## PROBLEM 3

```python
track_mean = df.groupby("Track")["Average"].mean().sort_values(ascending=False)
gender_mean = df.groupby("Gender")["Average"].mean().sort_values(ascending=False)
hometown_mean = df.groupby("Hometown")["Average"].mean().sort_values(ascending=False)

print("Mean Average by Track")
display(track_mean)

print("Mean Average by Gender")
display(gender_mean)

print("Mean Average by Hometown")
display(hometown_mean)
```

```python
track_mean = df.groupby("Track")["Average"].mean().sort_values(ascending=False)
gender_mean = df.groupby("Gender")["Average"].mean().sort_values(ascending=False)
hometown_mean = df.groupby("Hometown")["Average"].mean().sort_values(ascending=False)

print("Mean Average by Track")
display(track_mean)

print("Mean Average by Gender")
display(gender_mean)

print("Mean Average by Hometown")
display(hometown_mean)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].bar(track_mean.index, track_mean.values, color="steelblue")
axes[0].set_title("Mean Average by Track")
axes[0].set_xlabel("Track")
axes[0].set_ylabel("Mean Average")
axes[0].tick_params(axis="x", rotation=15)

axes[1].bar(gender_mean.index, gender_mean.values, color="seagreen")
axes[1].set_title("Mean Average by Gender")
axes[1].set_xlabel("Gender")
axes[1].set_ylabel("Mean Average")

axes[2].bar(hometown_mean.index, hometown_mean.values, color="indianred")
axes[2].set_title("Mean Average by Hometown")
axes[2].set_xlabel("Hometown")
axes[2].set_ylabel("Mean Average")

plt.tight_layout()
plt.show()

print("1. Track: Communication has the highest sample mean Average.")
print("2. Gender: Male has the highest sample mean Average.")
print("3. Hometown: Visayas has the highest sample mean Average.")
print("Note: These describe the observed dataset only. A difference in group means")
print("does not, by itself, establish that a feature causes a higher board-exam score.")
```
